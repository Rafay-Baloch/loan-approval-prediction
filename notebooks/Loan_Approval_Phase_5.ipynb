```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

df = pd.read_csv("cleaned_loan_approval_dataset.csv")

print("Dataset Loaded Successfully")
print("Dataset Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())
print("Target Variable: Loan_Status")
print("\nFirst Five Records:")

df.head()
```

## 2. Feature and Target Selection

The input features are separated from the target variable before model training.

```python
X = df.drop("Loan_Status", axis=1)
y = df["Loan_Status"]

print("FEATURE AND TARGET SELECTION")
print("-" * 45)
print("Input Features Shape:", X.shape)
print("Target Variable Shape:", y.shape)
print("\nTarget Variable: Loan_Status")
print("\nInput Features:")

for number, feature in enumerate(X.columns, start=1):
    print(f"{number}. {feature}")

print("\nTarget Distribution:")
print(y.value_counts().sort_index())
```

## 3. Train and Test Split

The dataset is divided into 80% training data and 20% testing data using stratified sampling.


```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("TRAIN AND TEST SPLIT")
print("-" * 45)
print("Complete Dataset Records:", len(df))
print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))
print("\nX_train Shape:", X_train.shape)
print("X_test Shape:", X_test.shape)
print("y_train Shape:", y_train.shape)
print("y_test Shape:", y_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts().sort_index())

print("\nTesting Target Distribution:")
print(y_test.value_counts().sort_index())
```

## 4. Logistic Regression Model Training

A Logistic Regression model is trained using the training data. StandardScaler is included to scale the input features.

```python
model = Pipeline([
    ("scaler", StandardScaler()),
    ("logistic_regression", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

model.fit(X_train, y_train)

training_accuracy = model.score(X_train, y_train)
testing_accuracy = model.score(X_test, y_test)

print("LOGISTIC REGRESSION MODEL TRAINING")
print("-" * 50)
print("Selected Algorithm: Logistic Regression")
print("Problem Type: Binary Classification")
print("Feature Scaling: StandardScaler")
print("Training Records Used:", len(X_train))
print("Number of Input Features:", X_train.shape[1])
print("Model Training Status: Successfully Trained")
print(f"Basic Training Accuracy: {training_accuracy * 100:.2f}%")
print(f"Basic Testing Accuracy: {testing_accuracy * 100:.2f}%")
```

## 5. Predictions and Prediction Probabilities

The trained model predicts loan approval classes and their probabilities for the testing data.

```python
y_pred = model.predict(X_test)
y_probability = model.predict_proba(X_test)

prediction_results = pd.DataFrame({
    "Actual_Status": y_test.to_numpy(),
    "Predicted_Status": y_pred,
    "Rejection_Probability": y_probability[:, 0],
    "Approval_Probability": y_probability[:, 1]
})

prediction_results["Rejection_Probability"] = (
    prediction_results["Rejection_Probability"] * 100
).round(2)

prediction_results["Approval_Probability"] = (
    prediction_results["Approval_Probability"] * 100
).round(2)

prediction_results["Actual_Result"] = prediction_results[
    "Actual_Status"
].map({
    0: "Rejected",
    1: "Approved"
})

prediction_results["Predicted_Result"] = prediction_results[
    "Predicted_Status"
].map({
    0: "Rejected",
    1: "Approved"
})

print("PREDICTIONS AND PROBABILITIES")
print("-" * 65)
print("0 = Rejected, 1 = Approved")
print("Decision Threshold: 50%")
print("\nFirst 10 Test Predictions:")

prediction_results[[
    "Actual_Result",
    "Predicted_Result",
    "Rejection_Probability",
    "Approval_Probability"
]].head(10)
```

## 6. Save the Trained Model

The complete preprocessing and Logistic Regression pipeline is saved for future predictions and application development.

```python
import joblib

joblib.dump(
    model,
    "loan_approval_logistic_model.pkl"
)

print("MODEL SAVING RESULT")
print("-" * 40)
print("Model saved successfully.")
print("File Name: loan_approval_logistic_model.pkl")
print("The saved pipeline includes StandardScaler and Logistic Regression.")
```
