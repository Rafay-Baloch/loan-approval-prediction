```python
import pandas as pd
import numpy as np

df = pd.read_csv("train_u6lujuX_CVtuZ9i.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
print("Missing Values Before Cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows Before Cleaning:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate Rows After Cleaning:", df.duplicated().sum())
```

```python
print("Data Types:")
print(df.dtypes)

categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{column}: {df[column].unique()}")
```

```python
categorical_missing = [
    "Gender",
    "Married",
    "Dependents",
    "Self_Employed",
    "Credit_History"
]

for column in categorical_missing:
    df[column] = df[column].fillna(df[column].mode()[0])

numerical_missing = [
    "LoanAmount",
    "Loan_Amount_Term"
]

for column in numerical_missing:
    df[column] = df[column].fillna(df[column].median())

print("Missing Values After Cleaning:")
print(df.isnull().sum())
```

```python
df = df.drop(columns=["Loan_ID"])

print("Loan_ID Removed Successfully")
print(df.columns.tolist())
```

```python
df["Dependents"] = df["Dependents"].replace({"3+": "3"}).astype(int)

binary_mapping = {
    "Gender": {"Male": 1, "Female": 0},
    "Married": {"Yes": 1, "No": 0},
    "Education": {"Graduate": 1, "Not Graduate": 0},
    "Self_Employed": {"Yes": 1, "No": 0},
    "Loan_Status": {"Y": 1, "N": 0}
}

for column, mapping in binary_mapping.items():
    df[column] = df[column].map(mapping)

df = pd.get_dummies(
    df,
    columns=["Property_Area"],
    dtype=int
)

print("Dataset After Encoding:")
df.head()
```

```python
encoded_columns = [
    "Gender",
    "Married",
    "Dependents",
    "Education",
    "Self_Employed",
    "Loan_Status",
    "Property_Area_Rural",
    "Property_Area_Semiurban",
    "Property_Area_Urban"
]

print("Encoded Dataset Preview:")
print(df[encoded_columns].head().to_string(index=False))
```

```python
print("Final Dataset Shape:", df.shape)
print("\nRemaining Missing Values:", df.isnull().sum().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
print("\nFinal Data Types:")
print(df.dtypes)

df.head()
```

```python
df.to_csv("cleaned_loan_approval_dataset.csv", index=False)

print("Cleaned dataset saved successfully.")
```
